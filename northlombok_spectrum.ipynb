```python
import numpy as np
import matplotlib.pyplot as plt

# Definisi fungsi sinc
def sinc(x):
    return np.sinc(x / np.pi)

# Fungsi untuk menghitung kontribusi dalam arah x
def s_x(x, z, dx):
    M, N = z.shape
    S_x = np.zeros_like(x, dtype=np.float64)
    
    for i in range(M):
        S_x += np.sum(z[i, :]) * sinc(np.pi * dx * (x - i / dx))
    
    return S_x

# Fungsi untuk menghitung kontribusi dalam arah y
def s_y(y, z, dy):
    M, N = z.shape
    S_y = np.zeros_like(y, dtype=np.float64)
    
    for j in range(N):
        S_y += np.sum(z[:, j]) * sinc(np.pi * dy * (y - j / dy))
    
    return S_y

# Fungsi untuk menghitung s(x, y) menggunakan kontribusi dari arah x dan y
def s(x, y, z, dx, dy):
    S_x = s_x(x, z, dx)
    S_y = s_y(y, z, dy)
    
    # Buat meshgrid untuk x dan y
    X, Y = np.meshgrid(x, y)
    
    # Kombinasikan kontribusi dari x dan y
    S = np.outer(S_y, S_x)
    
    return S

# Contoh penggunaan
# Misalkan kita memiliki data grid z_ij dan nilai dx, dy
z = np.random.rand(10, 10)  # Data grid z_ij
dx = 1.0  # Interval sampel dalam sumbu x
dy = 1.0  # Interval sampel dalam sumbu y

# Grid x dan y
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

# Hitung s(x, y)
S = s(x, y, z, dx, dy)

# Plot hasilnya
X, Y = np.meshgrid(x, y)
plt.contourf(X, Y, S)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Rekonstruksi Sinyal 2D')
plt.colorbar(label='s(x, y)')
plt.show()
```
